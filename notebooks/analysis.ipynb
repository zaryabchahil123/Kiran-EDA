```python
from pathlib import Path\n\nimport matplotlib.pyplot as plt\nimport pandas as pd\nimport seaborn as sns\n\nDATA_DIR = Path('../data')\ncups = pd.read_csv(DATA_DIR / 'WorldCups.csv')\nmatches = pd.read_csv(DATA_DIR / 'WorldCupMatches.csv').dropna(how='all')\nplayers = pd.read_csv(DATA_DIR / 'WorldCupPlayers.csv')\n\nmatches = matches.drop_duplicates(subset='MatchID', keep='first')\ncups['Attendance'] = cups['Attendance'].astype(str).str.replace('.', '', regex=False).astype(int)\nmatches['Total Goals'] = matches['Home Team Goals'] + matches['Away Team Goals']\n\ncups.shape, matches.shape, players.shape
```
